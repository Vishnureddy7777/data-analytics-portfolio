### Join the Tables

````python
# Merge FX rates
sales_fx = sales.merge(
    exchange_rates,
    left_on=['Order Date', 'Currency Code'],
    right_on=['Date', 'Currency'],
    how='left'
)

# Merge master data
sales_full = (
    sales_fx
    .merge(customers, on='CustomerKey', how='left')
    .merge(products, on='ProductKey', how='left')
    .merge(stores, on='StoreKey', how='left')
)
````

* A consolidated, analysis-ready dataset (sales_full) containing transactional, customer, product, store, and currency information in a single table.

### Normalise the Revenue

````python
# Create USD-normalised revenue
sales_full['Revenue_USD'] = (
    sales_full['Quantity'] *
    sales_full['Unit Price USD'] *
    sales_full['Exchange']
)
````

* Revenue was converted to USD to enable consistent cross-country financial analysis.

The metric used to assess a store’s revenue-generating efficiency relative to its physical size is **Revenue per Square Meter**.

This metric measures how much revenue a store generates for each square meter of its physical space. A higher 'Revenue per Square Meter' indicates better efficiency in utilizing the store's physical footprint to drive sales.

````python
# Calculate total revenue for each store
store_revenue = sales_full.groupby('StoreKey')['Revenue_USD'].sum().reset_index()

# Merge with the stores dataframe to get 'Square Meters'
store_efficiency = store_revenue.merge(stores[['StoreKey', 'Square Meters']], on='StoreKey', how='left')

# Calculate Revenue per Square Meter
store_efficiency['Revenue_per_Square_Meter'] = store_efficiency['Revenue_USD'] / store_efficiency['Square Meters']

# Sort by Revenue per Square Meter in descending order
store_efficiency_sorted = store_efficiency.sort_values(by='Revenue_per_Square_Meter', ascending=False)

# Display the top 10 stores by Revenue per Square Meter with formatted currency
print("Store Revenue Efficiency (Revenue per Square Meter):")
display(store_efficiency_sorted[['StoreKey', 'Revenue_per_Square_Meter']].head(10).style.format({'Revenue_per_Square_Meter': '${:,.2f}'}))
````

#### Insights:

*   **High Efficiency, Potentially Smaller Stores:** Stores with very high 'Revenue per Square Meter' might be smaller in physical size but are effectively utilizing their space to generate significant revenue. This could indicate a highly optimized layout, prime location, or a product mix that resonates strongly with their local customer base.

#### Recommendations:

*   **Investigate High-Performing Stores:** Analyze the operational strategies, product assortments, and location advantages of the top-performing stores (those with the highest Revenue per Square Meter). ### Perform Store Age–Based Segmentation

* **Q3.3.1 and 3.3.2** - Calculate the age of the stores in 5 buckets ('0–5 years', '5–10 years', '10–15 years', '15–20 years', '20+ years'). **(3 marks)**
* **Q3.3.3** - Calculate the total revenue in each bucket (write a Python code). **(2 marks)**
* **Q3.3.4** - Calculate the average revenue per store in each bucket (write a Python code). **(2 marks)**
* **Q3.3.5** - What insights and recommendations can be derived from this metric? **(2 marks)**

**Note** - Give atleast 1 insight and 1 recommendation


````python
# Determine a reference date for calculating store age (e.g., the latest order date in the sales data)
reference_date = sales_full['Order Date'].max()

# Calculate store age in years
stores['Store_Age_Years'] = (reference_date - stores['Open Date']).dt.days / 365.25

# Define age buckets
age_bins = [0, 5, 10, 15, 20, np.inf]
age_labels = ['0-5 years', '5-10 years', '10-15 years', '15-20 years', '20+ years']
stores['Age_Bucket'] = pd.cut(stores['Store_Age_Years'], bins=age_bins, labels=age_labels, right=False)

# Merge store age buckets with sales_full dataframe
sales_full_with_age = sales_full.merge(stores[['StoreKey', 'Age_Bucket']], on='StoreKey', how='left')

# Calculate total revenue per age bucket
total_revenue_by_age_bucket = sales_full_with_age.groupby('Age_Bucket', observed=False)['Revenue_USD'].sum().reindex(age_labels)
print("\nTotal Revenue by Store Age Bucket:")
print(total_revenue_by_age_bucket.map('${:,.2f}'.format))

# Calculate average revenue per store in each age bucket
# First, calculate total revenue per store
revenue_per_store = sales_full_with_age.groupby(['StoreKey', 'Age_Bucket'], observed=False)['Revenue_USD'].sum().reset_index()

# Then, calculate the number of stores and sum of revenue per age bucket
summary_by_age_bucket = revenue_per_store.groupby('Age_Bucket', observed=False).agg(
    Total_Revenue=('Revenue_USD', 'sum'),
    Number_of_Stores=('StoreKey', 'nunique')
).reindex(age_labels)

# Calculate average revenue per store in each age bucket
summary_by_age_bucket['Average_Revenue_per_Store'] = summary_by_age_bucket['Total_Revenue'] / summary_by_age_bucket['Number_of_Stores']

print("\nAverage Revenue per Store by Age Bucket:")
print(summary_by_age_bucket['Average_Revenue_per_Store'].map('${:,.2f}'.format))
````

#### Insights:

*   **Peak Performance in Mid-Life Stores:** Stores in the '10-15 years' age bucket generate the highest total revenue ($28.88M),
then the highest average revenue per store ($931,795.38). This suggests that stores reach their optimal performance during this age range, likely benefiting from established customer bases, operational maturity, and brand recognition.

*   **Underperformance in Very Young and Older Stores:** '0-5 years' stores have significantly lower total and average revenue. This is expected as new stores build their presence. Stores in the '15-20 years' bucket also show a drop in average revenue compared to the '10-15 years' group, indicating a potential decline in efficiency or market relevance as they age further.

*   **Zero Revenue for '20+ years' Bucket:** The '20+ years' age bucket shows $0.00 total revenue, and consequently, a NaN for average revenue. This requires immediate investigation, as it could indicate data issues (e.g., no sales recorded for stores in this category, or misclassification) or that these stores have been decommissioned without being removed from the `stores` dataset.

#### Recommendations:

*   **Replicate Success of '10-15 Years' Stores:** Analyze the operational strategies, marketing efforts, product assortments, and customer engagement models of stores within the '10-15 years' age bracket. Identify best practices and develop a playbook to apply these learnings to younger stores ('0-5 years', '5-10 years') to accelerate their growth and efficiency.

*   **Strategic Review for '15-20 Years' Stores:** Investigate the reasons for the decline in average revenue for stores aged '15-20 years'. This could involve modernization initiatives, targeted marketing campaigns, renegotiation of leases, or even consideration for relocation or closure if performance is consistently poor and not recoverable.

*   **Investigate '20+ Years' Data Anomaly:** Prioritize investigating why the '20+ years' age bucket has zero revenue. Confirm if these stores are still operational, if sales data exists for them, or if there's an error in data classification or extraction. If they are indeed non-operational, ensure they are correctly marked or removed from active analysis to avoid misleading metrics.
